1. write a function add(a,b) and call it using positional arguments.

```python
def add (a,b):
    return a + b
## calling function using positional arguments
result = add(2,5)
print(result)
```

2. write a function multiply(x,y)and call it using keyword arguments.

```python
def mul (x,y):
    return x * y
## calling function using keyword arguments
result = mul(x = 5, y = 2)
print(result)
```

3. write a function student(name,age) and call it using positional and keyword arguments.

```python
def student(name,age):
    print("Name =",name)
    print("Age =",age)
## calling using positional arguments
print("positional argument is")
student("Neha",24)
## calling using keyword arguments
print("keyword argument is")
student(age=24,name ="Neha")
```
